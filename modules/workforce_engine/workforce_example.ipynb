```python
# EJEMPLO COMPLETO

# Definicion del problema

setup = ProblemSetup(
        mobile_days_off_count=1,
        fixed_day_off=None,
        allowed_entry_hours= allowed_entry_hours,
        max_overcoverage_tolerance=0.2,
        closing_hour=20,
    )

initial_demand = np.ones((24, 28), dtype=int)
initial_demand[0:8,:]=0
initial_demand[18:,:]=0

state = WorkforceState(
        residual_demand=initial_demand,
        remaining_stock=np.array([1, 1, 1], dtype=int),
        expansion_mode=False,
        current_modality=None,
        current_entry_hour=None,
        assignment_week=0,
        initial_demand_total=int(initial_demand.sum()),
    )

engine = WorkforceEngine(setup)

# -----------------------------------------------------------------------------------------

is_terminal = False
action_trace = [] 

while is_terminal==False:

    # Enmascaro las acciones legales
    priors = np.random.rand(55)
    # Mascara de legalidad
    priors_mask = engine.legal_mask(state, priors)

    # Elijo una accion
    # Inicializa el generador - Simularia el MCTS
    rng = np.random.default_rng()

    # Elige un solo índice basado en las probabilidades
    action_id = rng.choice(len(priors_mask), p=priors_mask)
    
    # Aplico la accion
    result = engine.step(state, action_id)
    state = result.next_state
    is_terminal = result.is_terminal
    action_trace.append(action_id)


print("Trayectoria:", action_trace)
print("--------------------------------------------------------------------------------------")
print("Results", np.sum(state.residual_demand[state.residual_demand > 0]))
```
